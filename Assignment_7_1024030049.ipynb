```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage import data, util

def load_image(path=''):
    if not path.strip():
        return data.camera()
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(path)
    return img

def show(images, titles):
    fig, axes = plt.subplots(1, len(images), figsize=(4*len(images), 4), squeeze=False)
    for ax, im, title in zip(axes.ravel(), images, titles):
        ax.imshow(im, cmap='gray', vmin=0, vmax=255)
        ax.set_title(title)
        ax.axis('off')
    plt.tight_layout()
    plt.show()

img = load_image(input('Image path (Enter for sample): '))
```

```python
# 1
f = img.astype(np.float32)
lap = cv2.Laplacian(f,cv2.CV_32F,ksize=1)
sharp_lap = np.clip(f-lap,0,255).astype(np.uint8)
show([img,sharp_lap],['Original','Laplacian sharpening'])
```

```python
# 2
sx = cv2.Sobel(f,cv2.CV_32F,1,0,ksize=3)
sy = cv2.Sobel(f,cv2.CV_32F,0,1,ksize=3)
grad = cv2.magnitude(sx,sy)
grad = cv2.normalize(grad,None,0,255,cv2.NORM_MINMAX)
canny = cv2.Canny(img,100,200)
sharp_sobel = np.clip(f+0.5*grad,0,255).astype(np.uint8)
sharp_canny = np.clip(f+0.5*canny,0,255).astype(np.uint8)
show([img,sharp_sobel,sharp_canny],['Original','Sobel sharpening','Canny sharpening'])
```

```python
# 3
blur = cv2.GaussianBlur(f,(5,5),3)
mask = f-blur
sharp = np.clip(f+mask,0,255).astype(np.uint8)
show([img,blur,sharp],['Original','Gaussian blur','Unsharp masking'])
```
