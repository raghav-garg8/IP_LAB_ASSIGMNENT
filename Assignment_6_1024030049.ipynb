```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage import data, util

def load_image(path=''):
    if not path.strip():
        return data.camera()
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(path)
    return img

def show(images, titles):
    fig, axes = plt.subplots(1, len(images), figsize=(4*len(images), 4), squeeze=False)
    for ax, im, title in zip(axes.ravel(), images, titles):
        ax.imshow(im, cmap='gray', vmin=0, vmax=255)
        ax.set_title(title)
        ax.axis('off')
    plt.tight_layout()
    plt.show()

img = load_image(input('Image path (Enter for sample): '))
```

```python
k = int(input('Odd filter size (>=3): '))
sigma = float(input('Gaussian sigma: '))
assert k >= 3 and k%2 == 1 and sigma > 0
padding = input('Padding (zero/replication/reflect): ').strip().lower()
border = {'zero':cv2.BORDER_CONSTANT,'replication':cv2.BORDER_REPLICATE,'reflect':cv2.BORDER_REFLECT}[padding]
r = np.arange(-(k//2),k//2+1)
x,y = np.meshgrid(r,r)
avg = np.ones((k,k))/k**2
# Interpret the given separator as a maximum over |x| and |y|.
weighted = np.maximum(np.abs(x),np.abs(y)).astype(float)
weighted /= weighted.sum()
gaussian = np.exp(-(x*x+y*y)/(2*sigma*sigma))
gaussian /= gaussian.sum()
results = [cv2.filter2D(img,-1,h,borderType=border) for h in (avg,weighted,gaussian)]
show([img,*results],['Original','Averaging','Weighted','Gaussian'])
```
